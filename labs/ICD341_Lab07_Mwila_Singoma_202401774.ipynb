**QUESTION 1**: Use gradient descent to minimize 𝑓(𝑥) = (𝑥 − 10)^2.


## Question 1: Use Gradient Descent to Minimize \(f(x)=(x-10)^2\)

Gradient descent is an optimization algorithm that repeatedly updates a parameter in the direction that reduces the value of a function.

The update rule is:

$$
x_{\text{new}}=x_{\text{old}}-\alpha f'(x)
$$

where:

* \(x\) is the current value.
* \(\alpha\) is the learning rate.
* \(f'(x)\) is the derivative of the function.

For the function

$$
f(x)=(x-10)^2
$$

the derivative is

$$
f'(x)=2(x-10)
$$

We will use this derivative to iteratively find the minimum.


```python
# Lab 7 - Question 1: Gradient Descent

# Define the function
def f(x):
    return (x - 10)**2

# Define the derivative
def df(x):
    return 2 * (x - 10)

# Initialize parameters
x = 0.0
learning_rate = 0.1
iterations = 100

# Perform gradient descent
for i in range(iterations):
    x = x - learning_rate * df(x)

# Display results
print("Estimated minimum x:", x)
print("Minimum function value:", f(x))
```

**QUESTION 2**: Compare BFGS and CG on 𝑓(𝑥, 𝑦) = (𝑥 − 4)^2 + (𝑦 − 6)^2

## Question 2: Compare BFGS and Conjugate Gradient (CG)

BFGS and Conjugate Gradient (CG) are optimization algorithms used to find the minimum of a function.

* **BFGS:** Uses gradient information to build an approximation of the inverse Hessian matrix.
* **Conjugate Gradient (CG):** Uses gradient information and conjugate search directions to approach the minimum, generally without storing a full Hessian approximation.

The function is:

$$
f(x,y)=(x-4)^2+(y-6)^2
$$

The minimum occurs at \(x=4\) and \(y=6\), where the function value is zero.

We will use SciPy to apply both optimization methods and compare their results.


```python
# Lab 7 - Question 2: Compare BFGS and CG

import numpy as np
from scipy.optimize import minimize

# Define the function
def f(variables):
    x, y = variables
    return (x - 4)**2 + (y - 6)**2

# Initial guess
initial_guess = np.array([0.0, 0.0])

# Minimize using BFGS
result_bfgs = minimize(
    f,
    initial_guess,
    method="BFGS"
)

# Minimize using Conjugate Gradient
result_cg = minimize(
    f,
    initial_guess,
    method="CG"
)

# Display BFGS results
print("BFGS Results:")
print("Minimum point:", result_bfgs.x)
print("Minimum function value:", result_bfgs.fun)
print("Converged successfully:", result_bfgs.success)
print("Iterations:", result_bfgs.nit)
print("Function evaluations:", result_bfgs.nfev)
print("Gradient evaluations:", result_bfgs.njev)

# Display CG results
print("\nConjugate Gradient Results:")
print("Minimum point:", result_cg.x)
print("Minimum function value:", result_cg.fun)
print("Converged successfully:", result_cg.success)
print("Iterations:", result_cg.nit)
print("Function evaluations:", result_cg.nfev)
print("Gradient evaluations:", result_cg.njev)
```

### Results and Comparison

Both BFGS and Conjugate Gradient (CG) successfully minimized

$$
f(x,y)=(x-4)^2+(y-6)^2
$$

whose theoretical minimum is at \(x=4,\ y=6\) with \(f=0\). Starting from \((0,0)\):

| Quantity | BFGS | CG |
| --- | --- | --- |
| Estimated minimum point | (4.000002, 5.999999) | (4.000000, 6.000000) |
| Minimum function value | \(\approx 4.6\times10^{-12}\) | \(\approx 3.7\times10^{-16}\) |
| Iterations | 3 | 3 |
| Function evaluations | 12 | 21 |
| Gradient evaluations | 4 | 7 |
| Converged | Yes | Yes |

**Comparison:** Both methods reached the minimum with high accuracy in the same number of iterations. CG reached a slightly smaller function value in this run, while BFGS needed fewer function and gradient evaluations. Because no analytical gradient was supplied, SciPy approximated gradients numerically, which is why the evaluation counts are higher than the iteration counts. BFGS builds an approximation to the inverse Hessian, which usually gives faster convergence but needs \(O(n^2)\) memory. CG stores only a few vectors, so it is better suited to problems with very many variables. For this simple two-variable quadratic function, neither method has a meaningful advantage, and results on other functions may differ.

**Conclusion:** BFGS and Conjugate Gradient both found the minimum at approximately \((4,6)\) with a function value of approximately zero.

**QUESTION 3**: Explain the effect of changing the learning rate.

## Question 3: Explain the Effect of Changing the Learning Rate

The learning rate controls the size of each update during gradient descent.

The update rule is:

$$
x_{\text{new}}=x_{\text{old}}-\alpha f'(x)
$$

where \(\alpha\) represents the learning rate.

* **Small learning rate:** Updates are small, so convergence may be slow.
* **Appropriate learning rate:** The algorithm can approach the minimum efficiently.
* **Large learning rate:** Updates may overshoot the minimum, causing oscillation or divergence.

The learning rate must be chosen carefully to balance convergence speed and stability.


```python
# Lab 7 - Question 3: Effect of Learning Rate

def f(x):
    return (x - 10)**2

def df(x):
    return 2 * (x - 10)

learning_rates = [0.01, 0.1, 0.9, 1.1]
iterations = 20

for lr in learning_rates:
    x = 0.0

    for i in range(iterations):
        x = x - lr * df(x)

    print(f"Learning rate: {lr}")
    print(f"Estimated x: {x:.6f}")
    print(f"Function value: {f(x):.6f}")
    print()
```

### Results and Conclusion

The experiment demonstrates that the learning rate significantly affects the performance of gradient descent.

1. **Learning rate 0.01:** The algorithm converged slowly and remained far from the minimum after 20 iterations.
2. **Learning rate 0.1:** The algorithm approached the minimum steadily, obtaining an estimated value of \(x=9.884708\).
3. **Learning rate 0.9:** The algorithm also approached the minimum. It alternated across the minimum, but the distance decreased at the same rate as for a learning rate of 0.1.
4. **Learning rate 1.1:** The algorithm diverged, producing an estimated value of \(x=-373.375999\) and a very large function value.

**Conclusion:** A small learning rate can make convergence slow, while an excessively large learning rate can cause divergence. An appropriate learning rate helps gradient descent converge efficiently and reliably.


**QUESTION 4**: Use SciPy to estimate the mean of normally distributed observations using MLE.

## Question 4: Use SciPy to Estimate the Mean of Normally Distributed Observations Using MLE

Maximum Likelihood Estimation (MLE) is a statistical technique used to estimate unknown parameters by finding the values that maximize the likelihood of observing the available data.

For a normal distribution, when both the mean and variance are unknown, the maximum likelihood estimates are:

$$
\hat{\mu}=\frac{1}{n}\sum_{i=1}^{n}x_i
$$

and

$$
\hat{\sigma}^2=\frac{1}{n}\sum_{i=1}^{n}(x_i-\hat{\mu})^2
$$

The estimated mean is the sample mean. The variance formula uses \(n\), rather than \(n-1\), because this is the maximum likelihood estimator.


```python
# Lab 7 - Question 4: Maximum Likelihood Estimation

import numpy as np
from scipy.stats import norm

# Generate normally distributed observations
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)

# Estimate mean and standard deviation using MLE
estimated_mean, estimated_std = norm.fit(data)

# Display results
print("True population mean:", 50)
print("Estimated mean (MLE):", estimated_mean)
print("Estimated standard deviation (MLE):", estimated_std)
```

```python
# Verify by maximising the log-likelihood directly with scipy.optimize.minimize
from scipy.optimize import minimize

def neg_log_likelihood(params):
    mu, log_sigma = params
    sigma = np.exp(log_sigma)          # keeps sigma positive
    return -np.sum(norm.logpdf(data, loc=mu, scale=sigma))

result = minimize(neg_log_likelihood, x0=[0.0, 0.0], method="BFGS")
mu_hat, sigma_hat = result.x[0], np.exp(result.x[1])

print("Direct MLE using minimize:")
print("Estimated mean:", mu_hat)
print("Estimated standard deviation:", sigma_hat)
print("Sample mean (closed-form MLE):", np.mean(data))
print("Agrees with norm.fit():", np.allclose([mu_hat, sigma_hat], [estimated_mean, estimated_std], atol=1e-4))
```

### Results and Conclusion

The generated sample contained 1,000 observations from a normal distribution with a population mean of 50 and a standard deviation of 10.

Two approaches were used:

1. SciPy's `norm.fit()`, which returns the maximum likelihood estimates directly.
2. Minimising the negative log-likelihood with `scipy.optimize.minimize`. The standard deviation was parameterised as \(\sigma=e^{s}\) to keep it positive.

Both gave the same estimates: a mean of about **50.19** and a standard deviation of about **9.79**. The estimated mean equals the sample mean, as the theory predicts for the normal-distribution MLE. The estimates are close to the true values of 50 and 10, and the small differences arise from random sampling variation.

**Conclusion:** Maximum Likelihood Estimation estimates the parameters of a distribution by finding the values that make the observed data most probable. For the normal distribution, the MLE of the mean is the sample mean, and the optimisation approach confirms this.

**QUESTION 5**: Implement a sigmoid function.

## Question 5: Implement the Sigmoid Function

The sigmoid function is a mathematical function commonly used in machine learning to transform real-valued inputs into values between 0 and 1.

Its formula is:

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

The sigmoid function is useful in logistic regression because its output can be interpreted as a probability for binary classification.


```python
# Lab 7 - Question 5: Implement the Sigmoid Function

import numpy as np

# Define the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Test the function
values = np.array([-10, -2, 0, 2, 10])

print("Input values:")
print(values)

print("\nSigmoid outputs:")
print(sigmoid(values))
```

### Results and Conclusion

The sigmoid function was implemented successfully using NumPy.

The results show that:

1. Large negative inputs produce outputs close to 0.
2. An input of 0 produces an output of 0.5.
3. Large positive inputs produce outputs close to 1.
4. All outputs lie between 0 and 1.

**Conclusion:** The sigmoid function transforms real-valued inputs into values between 0 and 1. This property makes it useful in logistic regression for modelling probabilities in binary classification problems.

Fit a simple logistic regression model using MLE.

## Question 6: Logistic Regression Using Maximum Likelihood Estimation

Logistic regression is a statistical method used to model the probability of a binary outcome.

The logistic regression model is:

$$
P(Y=1\mid X)=\frac{1}{1+e^{-(\beta_0+\beta_1X)}}
$$

where:

* \(P(Y=1\mid X)\) is the probability of the outcome being 1.
* \(X\) represents study hours.
* \(\beta_0\) is the intercept.
* \(\beta_1\) is the coefficient for study hours.

Maximum Likelihood Estimation (MLE) estimates the coefficients by finding the values that maximize the likelihood of observing the given outcomes.

We will use SciPy's optimization function to estimate the coefficients and calculate predicted probabilities.


```python
# Lab 7 - Question 6: Logistic Regression Using MLE

import numpy as np
from scipy.optimize import minimize
from scipy.special import expit

# Define the data
X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
Y = np.array([0, 0, 0, 0, 1, 1, 1, 1], dtype=float)

# Define the negative log-likelihood function
def negative_log_likelihood(beta):
    beta0, beta1 = beta

    # Calculate predicted probabilities
    probabilities = expit(beta0 + beta1 * X)

    # Avoid log(0) using a small numerical constant
    epsilon = 1e-12
    probabilities = np.clip(probabilities, epsilon, 1 - epsilon)

    # Calculate negative log-likelihood
    log_likelihood = np.sum(
        Y * np.log(probabilities)
        + (1 - Y) * np.log(1 - probabilities)
    )

    return -log_likelihood

# Initial coefficient estimates
initial_beta = np.array([0.0, 0.0])

# Estimate coefficients using MLE
result = minimize(
    negative_log_likelihood,
    initial_beta,
    method="BFGS"
)

# Display the results
beta0, beta1 = result.x

print("Intercept (beta0):", beta0)
print("Study hours coefficient (beta1):", beta1)
print("Optimization successful:", result.success)

# Calculate predicted probabilities
predicted_probabilities = expit(beta0 + beta1 * X)

print("\nPredicted probabilities:")
for hours, probability in zip(X, predicted_probabilities):
    print(f"{hours:.0f} study hours: {probability:.4f}")
```

```python
# Demonstrate complete separation: scaling the coefficients keeps improving the likelihood
beta_hat = result.x
print("Decision boundary (where P = 0.5): x =", -beta_hat[0] / beta_hat[1])
print()
print("Negative log-likelihood for scaled versions of the fitted coefficients:")
for k in [0.25, 0.5, 1, 2]:
    print(f"  scale {k:>4}: {negative_log_likelihood(k * beta_hat):.6f}")
```

### Results and Conclusion

Logistic regression was fitted by minimising the negative log-likelihood with SciPy. The optimiser returned approximately

* Intercept: \(\beta_0\approx-127.1\)
* Study hours coefficient: \(\beta_1\approx28.03\)

The positive coefficient indicates that the probability of passing increases with study hours. The fitted probabilities are essentially 0 for 1 to 4 hours and essentially 1 for 5 to 8 hours. The estimated decision boundary (where the probability is 0.5) is at about **4.5 hours**, which lies between the last failing student (4 hours) and the first passing student (5 hours).

**Statistical limitation (complete separation):** The two outcome classes are perfectly separated by study hours, so the likelihood keeps increasing as the coefficients grow larger in the same direction. The scaling experiment shows this: multiplying the fitted coefficients by a larger factor keeps lowering the negative log-likelihood towards zero. Consequently, a finite maximum-likelihood estimate does not exist, and the values of \(\beta_0\) and \(\beta_1\) above depend on where the optimiser stopped, so they should not be interpreted literally. Only the direction of the effect and the decision boundary of about 4.5 hours are meaningful. In practice, this problem is addressed with a penalised (regularised) logistic regression or by collecting more data.

**Conclusion:** The experiment demonstrates how logistic regression combines the sigmoid function with maximum likelihood estimation, and how complete separation prevents finite maximum-likelihood estimates from existing.
