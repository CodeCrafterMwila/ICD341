### Question 1: Implement a Median Function

The median is the middle value of a dataset after the values have been
sorted.

- If the dataset contains an odd number of values, the median is the
  middle value.
- If the dataset contains an even number of values, the median is the
  average of the two middle values.
- If the input is empty, the function should raise a ValueError.

The function will be tested using a dataset with a known median and an
empty input to verify that it handles invalid input correctly.

```python
def calculate_median(data):
    """
    Calculate the median of a list of numbers.

    Raises:
        ValueError: If the input data is empty.
    """
    if len(data) == 0:
        raise ValueError("Cannot calculate the median of an empty dataset.")

    sorted_data = sorted(data)
    n = len(sorted_data)
    middle = n // 2

    # Odd number of values
    if n % 2 == 1:
        return sorted_data[middle]

    # Even number of values
    return (sorted_data[middle - 1] + sorted_data[middle]) / 2
```

```python
# Test 1: Dataset with an odd number of values
data = [7, 2, 9, 4, 5]
result = calculate_median(data)

print("Dataset: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:5", data)
print("Calculated median: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:6", result)

assert result == 5
print("Test 1 passed! - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:9")
```

```python
# Test 2: Dataset with an even number of values
data = [1, 2, 3, 4]
result = calculate_median(data)

print("Dataset: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:5", data)
print("Calculated median: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:6", result)

assert result == 2.5
print("Test 2 passed! - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:9")
```

```python
# Test 3: Empty dataset
try:
    calculate_median([])
except ValueError as error:
    print("Empty input handled correctly: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:5", error)
else:
    raise AssertionError("The function should raise a ValueError.")
```

### Interpretation

The median function successfully calculates the middle value for datasets
containing both odd and even numbers of observations.

The tests confirm that:
- The median of an odd-sized dataset is the middle sorted value.
- The median of an even-sized dataset is the average of the two middle values.
- An empty dataset raises a ValueError instead of producing an invalid result.

All tests passed, demonstrating that the function behaves correctly for
the tested cases.

## Question 2: Testing a Probability Estimate

A probability estimate represents the likelihood of an event occurring.

A valid probability must satisfy:

0 ≤ P(A) ≤ 1

We can estimate a probability by dividing the number of successful
outcomes by the total number of trials.

To test our function, we will use Python assertions to verify that
the calculated probability always falls between 0 and 1.

```python
def estimate_probability(successes, trials):
    """
    Estimate probability using successes divided by total trials.
    """

    if trials <= 0:
        raise ValueError("Number of trials must be greater than zero.")

    if successes < 0 or successes > trials:
        raise ValueError("Successes must be between 0 and trials.")

    probability = successes / trials

    return probability
```

```python
# Test 1: Normal probability estimate
p = estimate_probability(8, 12)

print("Estimated probability: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:4", p)

assert 0 <= p <= 1

print("Test 1 passed: Probability is within [0, 1]. - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:8")
```

```python
# Test 2: Probability equals zero
p_zero = estimate_probability(0, 10)

assert 0 <= p_zero <= 1

print("Probability zero: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:6", p_zero)

# Test 3: Probability equals one
p_one = estimate_probability(10, 10)

assert 0 <= p_one <= 1

print("Probability one: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:13", p_one)

print("Boundary tests passed! - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:15")
```

```python
# Test 4: Invalid number of successes
try:
    estimate_probability(15, 10)
except ValueError as error:
    print("Invalid input handled correctly: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:5", error)
```

### Interpretation

The probability estimation function successfully calculates probabilities
using the ratio of successful outcomes to total trials.

The tests confirm that:

1. A normal probability estimate falls between 0 and 1.
2. A probability can equal 0 when no successes occur.
3. A probability can equal 1 when every trial is successful.
4. Invalid inputs are rejected using ValueError.

The assertions confirm that the estimated probabilities satisfy the
mathematical condition 0 ≤ P(A) ≤ 1.

## Question 3: Implement a Transformer for Column Scaling

A transformer is an object that modifies or transforms data into a
different representation.

In this question, we implement a transformer that scales each column
of a dataset by its standard deviation.

The standard deviation measures how spread out the values in a dataset
are around their mean.

For each column, the transformation is:

X_scaled = X / standard_deviation

The transformer will:

1. Calculate the standard deviation of each column.
2. Divide each column by its corresponding standard deviation.
3. Return the scaled dataset.

This helps place numerical features on comparable scales.

```python
import numpy as np
```

```python
class StandardDeviationTransformer:

    def fit(self, X):
        """
        Calculate the standard deviation of each column.
        """
        self.std_ = np.std(X, axis=0)

        # Prevent division by zero
        self.std_[self.std_ == 0] = 1

        return self

    def transform(self, X):
        """
        Scale each column by its standard deviation.
        """
        return X / self.std_

    def fit_transform(self, X):
        """
        Fit the transformer and transform the data.
        """
        self.fit(X)
        return self.transform(X)
```

```python
# Create a sample dataset
data = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000],
    [40, 400, 4000],
    [50, 500, 5000]
])

print("Original Dataset: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:10")
print(data)
```

```python
# Create transformer object
transformer = StandardDeviationTransformer()

# Transform the dataset
scaled_data = transformer.fit_transform(data)

print("Column Standard Deviations: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:7")
print(transformer.std_)

print("\nScaled Dataset: - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:10")
print(scaled_data)
```

```python
# Verify the transformation
expected = data / np.std(data, axis=0)

assert np.allclose(scaled_data, expected)

print("Verification successful! - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:6")
print("All columns were correctly scaled. - ICD341_Lab12_Mwila_Singoma_202401774.ipynb:7")
```

### Interpretation

The StandardDeviationTransformer successfully scales each column of
the dataset by its corresponding standard deviation.

The standard deviations are calculated independently for each column
using NumPy's axis=0 parameter.

The transformed dataset contains values divided by their respective
column standard deviations.

The verification test using np.allclose() confirms that the
transformation was correctly implemented.

This demonstrates how a custom transformer can be used to perform
feature scaling in statistical computing.
