
```bash

```

## 2. Verify Python Installation

The following code displays the installed Python version and the location of the Python interpreter being used.

This confirms that the laboratory is running inside the intended `icd341` Conda environment.

````python
import sys
print(sys.version)
print(sys.executable)
````

### Understanding the Output

The `sys.version` attribute displays information about the Python version currently running the program. In this laboratory, it confirms that Python 3.12.14 is being used.

The `sys.executable` attribute displays the full path to the Python interpreter executing the code. In this case, the path confirms that the notebook is using the Python interpreter inside the `icd341` Conda environment.

Therefore, `sys.version` identifies the Python version, while `sys.executable` identifies the location of the Python interpreter.

## 4. Short-Answer Questions

### 1. What is a Python environment?

A Python environment is an isolated workspace containing a specific Python interpreter and the packages required for a project. It allows different projects to use different package versions without interfering with one another.

### 2. Why should we create separate environments for different courses or projects?

Separate environments prevent package-version conflicts. Different courses or projects may require different versions of Python libraries. Using isolated environments ensures that installing or updating a package for one project does not affect another project.

### 3. What is Conda?

Conda is an environment and package management tool. It allows users to create isolated Python environments, install packages, and manage dependencies.

### 4. How do you activate a Conda environment?

An environment can be activated using:

`conda activate icd341`

This activates the ICD341 environment so that Python and its installed packages can be used.

### 5. Why is specifying a Python version important?

Specifying a Python version helps maintain compatibility with required packages and project code. It also improves reproducibility by ensuring that the same Python version is used when recreating the environment.

### 6. What is JupyterLab?

JupyterLab is an interactive development environment that allows users to create and execute notebooks containing code, explanations, visualizations, and outputs. It is useful for scientific computing and data analysis.

### 7. Why should VS Code use the correct Python interpreter?

Selecting the correct Python interpreter ensures that VS Code executes code using the intended environment and has access to the packages installed in that environment.

### 8. What is Git and GitHub?

Git is a version-control system used to track changes in project files. GitHub is an online platform for hosting Git repositories, sharing code, and collaborating with other developers.

### 9. What is the difference between `git add` and `git commit`?

`git add` stages changes so they can be included in the next commit. `git commit` saves the staged changes as a snapshot in the repository's version history.

### 10. What is the purpose of a `.gitignore` file?

A `.gitignore` file specifies files and directories that Git should ignore. It helps prevent unnecessary files, temporary files, environment folders, and sensitive configuration files from being included in the repository.

### 11. Why should Jupyter checkpoint files be excluded from Git?

Jupyter checkpoint files are automatically generated backups of notebooks. They are usually unnecessary in a Git repository because Git already tracks notebook changes. Excluding them helps keep the repository clean and organised.

### 12. What is reproducibility in scientific computing?

Reproducibility means being able to repeat an experiment or computation and obtain consistent results using the same code, data, environment, and configuration.
