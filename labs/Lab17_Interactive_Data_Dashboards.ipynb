## Exercise 1: Building an Interactive Maize Production Dashboard

### Implementation

I developed an interactive dashboard using Streamlit, Pandas and Matplotlib.

The dashboard uses an illustrative sample dataset containing maize production figures for five selected Zambian districts. These figures are for educational purposes and do not represent official agricultural statistics.

The dashboard includes:

1. A title and introduction describing the application.
2. A Pandas DataFrame containing maize production data.
3. An interactive sidebar slider for filtering districts by minimum production.
4. Four statistical metrics displaying total, average, highest and lowest production.
5. A Matplotlib bar chart comparing production across districts, including value labels.
6. An interactive dropdown menu for selecting and exploring individual districts.
7. A CSV download button for exporting the filtered dataset.

### Results

Using the complete sample dataset:

- Total production: 2,300 tonnes.
- Average production: 460 tonnes.
- Highest production: Chipata (600 tonnes).
- Lowest production: Chongwe (350 tonnes).

The interactive filter was tested successfully. Selecting a minimum production of 500 tonnes displayed Lusaka and Chipata, while selecting 600 tonnes displayed only Chipata.

The CSV download feature also worked successfully.
