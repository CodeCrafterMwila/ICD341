## Exercise 1: Building an Interactive Maize Production Dashboard

### Implementation

I developed an interactive dashboard using Streamlit, Pandas and Matplotlib.

The dashboard includes:

1. A title and introduction describing the application.
2. A Pandas DataFrame containing maize production data for five Zambian districts.
3. Statistical summaries showing total and average production.
4. A Matplotlib bar chart comparing production across districts.
5. An interactive dropdown menu that allows users to select a district and view its production.

### Results

- Total maize production: 2,300 tonnes.
- Average maize production: 460 tonnes.
- Highest production: Chipata (600 tonnes).
- Lowest production: Chongwe (350 tonnes).
