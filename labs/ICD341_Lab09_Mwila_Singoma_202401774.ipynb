
## Lab 1: Bootstrap and Resampling

### Question 1: Bootstrap the Mean

**Objective:** To estimate the sampling distribution of the sample mean using bootstrap resampling and obtain a 95% percentile confidence interval.

Bootstrap resampling involves repeatedly drawing samples from the original dataset with replacement. Each bootstrap sample has the same size as the original dataset.

The 95% percentile confidence interval is obtained using the 2.5th and 97.5th percentiles of the bootstrap distribution.

```python

import numpy as np

# Original dataset
data = np.array([3, 4, 5, 7, 8, 9])

# Calculate the original sample mean
original_mean = np.mean(data)

print("Original sample mean:", original_mean)
```

```python

# Set the random seed for reproducibility
np.random.seed(42)

# Number of bootstrap resamples
n_bootstrap = 10000

# Generate bootstrap samples with replacement
bootstrap_samples = np.random.choice(
    data,
    size=(n_bootstrap, len(data)),
    replace=True
)

# Calculate the mean of each bootstrap sample
bootstrap_means = np.mean(bootstrap_samples, axis=1)

print("Number of bootstrap samples:", n_bootstrap)
print("First five bootstrap means:", bootstrap_means[:5])
```

```python

# Calculate the 95% percentile confidence interval
lower_bound = np.percentile(bootstrap_means, 2.5)
upper_bound = np.percentile(bootstrap_means, 97.5)

print("Bootstrap mean:", np.mean(bootstrap_means))
print("95% Percentile Confidence Interval:")
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

### Interpretation

The original sample mean is 6.0. Bootstrap resampling repeatedly selects six observations from the original dataset with replacement, producing 10,000 bootstrap samples.

The mean is calculated for each resample to create an approximate sampling distribution of the sample mean. The 2.5th and 97.5th percentiles of this distribution form the 95% percentile confidence interval.

This interval describes the uncertainty in the estimated population mean based on the available sample. A wider interval indicates greater uncertainty, while a narrower interval indicates greater precision.

### Question 2: Bootstrap the Median

**Objective:** To estimate the sampling distribution of the median using bootstrap resampling.

The median is a measure of central tendency that represents the middle value of a dataset.

Since the dataset contains an even number of observations, the median is calculated as the average of the two middle values.

Bootstrap resampling involves repeatedly sampling from the original dataset with replacement and calculating the median of each resample.

```python
# Original dataset
data = np.array([3, 4, 5, 7, 8, 9])

# Calculate the original median
original_median = np.median(data)

print("Original sample median:", original_median)
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Number of bootstrap samples
n_bootstrap = 10000

# Generate bootstrap samples with replacement
bootstrap_samples = np.random.choice(
    data,
    size=(n_bootstrap, len(data)),
    replace=True
)

# Calculate the median of each bootstrap sample
bootstrap_medians = np.median(bootstrap_samples, axis=1)

print("Number of bootstrap samples:", n_bootstrap)
print("First five bootstrap medians:", bootstrap_medians[:5])
```

```python
# Calculate the 95% percentile confidence interval
lower_bound = np.percentile(bootstrap_medians, 2.5)
upper_bound = np.percentile(bootstrap_medians, 97.5)

# Calculate the bootstrap estimate
bootstrap_median_estimate = np.mean(bootstrap_medians)

print("Original sample median:", original_median)
print("Bootstrap median estimate:", bootstrap_median_estimate)
print("95% Percentile Confidence Interval:")
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

### Interpretation

The original sample median is 6.0, calculated as the average of the two middle observations, 5 and 7.

The bootstrap procedure repeatedly draws samples with replacement from the original dataset and calculates the median of each sample. These 10,000 bootstrap medians form an approximate sampling distribution of the median.

The 2.5th and 97.5th percentiles provide the 95% percentile confidence interval. This interval helps describe the uncertainty associated with estimating the population median from the available observations.

The median is less sensitive to extreme observations than the mean, making it a useful measure of central tendency when data contain outliers.

### Question 3: Parametric Bootstrap from a Fitted Normal Model

**Objective:** To perform a parametric bootstrap by fitting a normal distribution to the original dataset and generating simulated samples from the fitted model.

A parametric bootstrap assumes that the data follow a specified probability distribution.

In this experiment, we assume the data follow a normal distribution:

X ~ N(μ, σ²)

The sample mean and standard deviation are estimated from the original data. New samples are then generated from the fitted normal distribution, and their means are calculated to estimate the sampling distribution.

```python
import numpy as np

# Original dataset
data = np.array([3, 4, 5, 7, 8, 9])

# Estimate parameters of the normal distribution
mu_hat = np.mean(data)
sigma_hat = np.std(data, ddof=1)

print("Estimated mean:", mu_hat)
print("Estimated standard deviation:", sigma_hat)
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Number of bootstrap samples
n_bootstrap = 10000

# Generate samples from the fitted normal distribution
parametric_samples = np.random.normal(
    loc=mu_hat,
    scale=sigma_hat,
    size=(n_bootstrap, len(data))
)

# Calculate the mean of each simulated sample
parametric_means = np.mean(parametric_samples, axis=1)

print("Number of simulated samples:", n_bootstrap)
print("First five simulated means:", parametric_means[:5])
```

```python
# Calculate the 95% percentile confidence interval
lower_bound = np.percentile(parametric_means, 2.5)
upper_bound = np.percentile(parametric_means, 97.5)

print("Parametric bootstrap mean:", np.mean(parametric_means))
print("95% Percentile Confidence Interval:")
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

### Interpretation

The original dataset is assumed to follow a normal distribution. Its parameters are estimated using the sample mean and sample standard deviation.

The parametric bootstrap generates 10,000 new samples from this fitted normal distribution, with each simulated sample having the same size as the original dataset.

The mean of each simulated sample is calculated to form an approximate sampling distribution. The 2.5th and 97.5th percentiles of these bootstrap means provide the 95% percentile confidence interval.

Unlike the nonparametric bootstrap, which resamples directly from the observed data, the parametric bootstrap generates new observations from an assumed probability distribution.

The reliability of this method depends on whether the normal distribution is a reasonable model for the underlying data.

### Question 4: Data Leakage During Model Validation

Data leakage occurs when information from outside the training dataset is unintentionally used during model training or validation. This can produce overly optimistic performance estimates because the model indirectly gains access to information it would not normally have when making predictions on unseen data.

**Example: Leakage during standardization**

Suppose a dataset is standardized using the mean and standard deviation calculated from the entire dataset before it is divided into training and testing sets.

This introduces data leakage because information from the test set influences the preprocessing parameters used during training.

**Prevention**

To prevent data leakage:

1. Split the dataset into training and testing sets before fitting preprocessing transformations.
2. Calculate preprocessing parameters, such as the mean and standard deviation, using only the training data.
3. Apply the learned transformations to validation and test data without fitting them again.
4. Use a scikit-learn Pipeline to ensure preprocessing is fitted correctly within each training fold during cross-validation.

**Conclusion**

Preventing data leakage is essential because it ensures that model validation provides a more realistic estimate of performance on genuinely unseen data.
