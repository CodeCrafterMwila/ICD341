
## Lab 1: Bootstrap and Resampling

### Question 1: Bootstrap the Mean

**Objective:** To estimate the sampling distribution of the sample mean using bootstrap resampling and obtain a 95% percentile confidence interval.

Bootstrap resampling involves repeatedly drawing samples from the original dataset with replacement. Each bootstrap sample has the same size as the original dataset.

The 95% percentile confidence interval is obtained using the 2.5th and 97.5th percentiles of the bootstrap distribution.

```python

import numpy as np

# Original dataset
data = np.array([3, 4, 5, 7, 8, 9])

# Calculate the original sample mean
original_mean = np.mean(data)

print("Original sample mean: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:10", original_mean)
```

```python

# Set the random seed for reproducibility
np.random.seed(42)

# Number of bootstrap resamples
n_bootstrap = 10000

# Generate bootstrap samples with replacement
bootstrap_samples = np.random.choice(
    data,
    size=(n_bootstrap, len(data)),
    replace=True
)

# Calculate the mean of each bootstrap sample
bootstrap_means = np.mean(bootstrap_samples, axis=1)

print("Number of bootstrap samples: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:18", n_bootstrap)
print("First five bootstrap means: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:19", bootstrap_means[:5])
```

```python

# Calculate the 95% percentile confidence interval
lower_bound = np.percentile(bootstrap_means, 2.5)
upper_bound = np.percentile(bootstrap_means, 97.5)

print("Bootstrap mean: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:6", np.mean(bootstrap_means))
print("95% Percentile Confidence Interval: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:7")
print("Lower bound: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:8", lower_bound)
print("Upper bound: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:9", upper_bound)
```

### Question 2: Bootstrap the Median

**Objective:** To estimate the sampling distribution of the median using bootstrap resampling.

The median is a measure of central tendency that represents the middle value of a dataset.

Since the dataset contains an even number of observations, the median is calculated as the average of the two middle values.

Bootstrap resampling involves repeatedly sampling from the original dataset with replacement and calculating the median of each resample.

```python
# Original dataset
data = np.array([3, 4, 5, 7, 8, 9])

# Calculate the original median
original_median = np.median(data)

print("Original sample median:", original_median)
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Number of bootstrap samples
n_bootstrap = 10000

# Generate bootstrap samples with replacement
bootstrap_samples = np.random.choice(
    data,
    size=(n_bootstrap, len(data)),
    replace=True
)

# Calculate the median of each bootstrap sample
bootstrap_medians = np.median(bootstrap_samples, axis=1)

print("Number of bootstrap samples: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:17", n_bootstrap)
print("First five bootstrap medians: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:18", bootstrap_medians[:5])
```

```python
# Calculate the 95% percentile confidence interval
lower_bound = np.percentile(bootstrap_medians, 2.5)
upper_bound = np.percentile(bootstrap_medians, 97.5)

# Calculate the bootstrap estimate
bootstrap_median_estimate = np.mean(bootstrap_medians)

print("Original sample median: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:8", original_median)
print("Bootstrap median estimate: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:9", bootstrap_median_estimate)
print("95% Percentile Confidence Interval: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:10")
print("Lower bound: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:11", lower_bound)
print("Upper bound: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:12", upper_bound)
```

### Question 3: Parametric Bootstrap from a Fitted Normal Model

**Objective:** To perform a parametric bootstrap by fitting a normal distribution to the original dataset and generating simulated samples from the fitted model.

A parametric bootstrap assumes that the data follow a specified probability distribution.

In this experiment, we assume the data follow a normal distribution:

X ~ N(μ, σ²)

The sample mean and standard deviation are estimated from the original data. New samples are then generated from the fitted normal distribution, and their means are calculated to estimate the sampling distribution.

```python
import numpy as np

# Original dataset
data = np.array([3, 4, 5, 7, 8, 9])

# Estimate parameters of the normal distribution
mu_hat = np.mean(data)
sigma_hat = np.std(data, ddof=1)

print("Estimated mean:", mu_hat)
print("Estimated standard deviation:", sigma_hat)
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Number of bootstrap samples
n_bootstrap = 10000

# Generate samples from the fitted normal distribution
parametric_samples = np.random.normal(
    loc=mu_hat,
    scale=sigma_hat,
    size=(n_bootstrap, len(data))
)

# Calculate the mean of each simulated sample
parametric_means = np.mean(parametric_samples, axis=1)

print("Number of simulated samples: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:17", n_bootstrap)
print("First five simulated means: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:18", parametric_means[:5])
```

```python
# Calculate the 95% percentile confidence interval
lower_bound = np.percentile(parametric_means, 2.5)
upper_bound = np.percentile(parametric_means, 97.5)

print("Parametric bootstrap mean: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:5", np.mean(parametric_means))
print("95% Percentile Confidence Interval: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:6")
print("Lower bound: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:7", lower_bound)
print("Upper bound: - ICD341_Lab09_Mwila_Singoma_202401774.ipynb:8", upper_bound)
```

### Question 4: Data Leakage During Model Validation

Data leakage occurs when information from the validation or test dataset unintentionally influences the training process, resulting in an overly optimistic evaluation of the model's performance.

**Example:**

Data leakage can occur when a dataset is standardized using the mean and standard deviation of the entire dataset before splitting it into training and testing sets.

In this case, information from the test dataset is used during preprocessing, allowing the model to indirectly access information about the unseen data.

**Prevention:**

To prevent data leakage, the dataset should first be divided into training and testing sets. Preprocessing parameters, such as the mean and standard deviation used for standardization, should be calculated using only the training data. The same parameters are then applied to the validation and test data.

**Conclusion:**

Data leakage must be avoided to ensure that model validation provides a reliable estimate of how well the model will perform on unseen data.
