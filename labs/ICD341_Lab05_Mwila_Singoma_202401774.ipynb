**Question 1**: Solve a system of linear equations using NumPy: Solve 3𝑥 + 2𝑦 = 12
 𝑥 + 4𝑦 = 10
Using NumPy.

```python
# Import NumPy
import numpy as np

# Define the coefficient matrix
A = np.array([
    [3, 2],
    [1, 4]
])

# Define the constants vector
b = np.array([12, 10])

# Solve the system of linear equations
solution = np.linalg.solve(A, b)

# Display the results
print("Solution:")
print(f"x = {solution[0]:.2f}")
print(f"y = {solution[1]:.2f}")
```

```python
# Verify the solution
verification = A @ solution

print("Original constants:", b)

print("Solution is correct:", np.allclose(verification, b))
```

**Conclusion**

The system of linear equations was successfully solved using NumPy's `np.linalg.solve()` function. The calculated values were \(x=2.8\) and \(y=1.8\). Substituting these values into the original equations confirmed that both equations were satisfied.


**Question 2**: Solving a System of Three Linear Equations: Create a 3 × 3 matrix and solve a system using **np.linalg.solve()**.

```python
# Import NumPy
import numpy as np

# Define the 3x3 coefficient matrix
A = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [3, 2, 4]
], dtype=float)

# Define the constants vector
b = np.array([10, 18, 25], dtype=float)

# Solve the system of equations
solution = np.linalg.solve(A, b)

# Display the solution
print("Coefficient matrix A:")
print(A)

print("\nConstants vector b:")
print(b)

print("\nSolution:")
print(f"x = {solution[0]:.4f}")
print(f"y = {solution[1]:.4f}")
print(f"z = {solution[2]:.4f}")
```

```python
# Verify the solution using matrix multiplication
verification = A @ solution

print("Original constants:", b)

# Check whether both sides agree
print("Solution is correct:", np.allclose(verification, b))
```

**Conclusion**

The system of three simultaneous linear equations was successfully solved using NumPy's `np.linalg.solve()` function. The resulting values were \(x \approx 1.7273\), \(y \approx 3.1818\), and \(z \approx 3.3636\). Matrix multiplication was used to verify the solution, confirming that the calculated values satisfy the original system of equations.


**Conclusion**

`np.linalg.solve(A, b)` is generally preferred to explicitly calculating \(A^{-1}b\) for three reasons:

1. **Efficiency:** `solve` uses an LU factorisation of \(A\) and then substitutes, which costs roughly \(\tfrac{2}{3}n^3\) operations. Computing the inverse requires solving \(n\) systems (one per column of the identity matrix) and then performing a further matrix-vector multiplication.
2. **Numerical accuracy:** Forming \(A^{-1}\) introduces extra rounding error. The Hilbert-matrix experiment above shows this: the solution obtained with `solve` had a much smaller error and a residual near machine precision (about \(10^{-16}\)), while the inverse-based solution had a residual several orders of magnitude larger.
3. **Directness:** `solve` addresses \(Ax=b\) directly, and the inverse is not needed to answer the question.

For a small well-conditioned system such as \(3x+2y=12,\ x+4y=10\), both methods give the same answer, but `solve` is the recommended method whenever the coefficient matrix is square and nonsingular.

```python
import numpy as np

# Define the coefficient matrix
A = np.array([
    [3, 2],
    [1, 4]
])

# Define the constants vector
b = np.array([12, 10])

# Method 1: Solve directly
solution1 = np.linalg.solve(A, b)

# Method 2: Use the inverse
solution2 = np.linalg.inv(A) @ b

# Display the results
print("Solution using np.linalg.solve():")
print(solution1)

print("\nSolution using matrix inverse:")
print(solution2)
```

For the small, well-conditioned system above the two methods agree. The difference appears when the matrix is **ill-conditioned**. The Hilbert matrix is a classic example. The true solution is chosen as a vector of ones, so \(b=Hx\) and the exact answer is known.

```python
from scipy.linalg import hilbert

n = 10
H = hilbert(n)              # ill-conditioned matrix
x_true = np.ones(n)         # known exact solution
b_h = H @ x_true

x_solve = np.linalg.solve(H, b_h)
x_inv = np.linalg.inv(H) @ b_h

print(f"Condition number of H: {np.linalg.cond(H):.3e}")
print()
print("Maximum error in the solution:")
print(f"  np.linalg.solve(H, b):  {np.abs(x_solve - x_true).max():.3e}")
print(f"  np.linalg.inv(H) @ b:   {np.abs(x_inv - x_true).max():.3e}")
print()
print("Residual norm ||Hx - b||:")
print(f"  np.linalg.solve(H, b):  {np.linalg.norm(H @ x_solve - b_h):.3e}")
print(f"  np.linalg.inv(H) @ b:   {np.linalg.norm(H @ x_inv - b_h):.3e}")
```

**Conclusion**

np.linalg.solve(A, b) is generally preferred to explicitly calculating \(A^{-1}b\) because it is computationally more efficient, avoids unnecessary matrix inversion, and typically provides better numerical accuracy. It directly solves the system of linear equations \(Ax=b\). Therefore, it is the recommended method when the coefficient matrix is square and nonsingular.

**QUESTION 4**: The data below relate to consumption expenditure and disposable income for a random 
sample of 10 households in Lusaka.
| Household | Disposable Income (X) | Consumption (Y) |
| --------: | --------------------: | --------------: |
|         1 |                    30 |              25 |
|         2 |                    45 |              20 |
|         3 |                    60 |              40 |
|         4 |                   100 |              65 |
|         5 |                   150 |             120 |
|         6 |                   200 |             190 |
|         7 |                   240 |             200 |
|         8 |                   250 |             200 |
|         9 |                   300 |             260 |
|        10 |                   320 |             220 |

(i) Use a regression design matrix to estimate regression coefficients.

(ii) Estimate the consumption for a household with disposable income of 181

```python
import numpy as np

# Disposable income (X)
income = np.array([30, 45, 60, 100, 150, 200, 240, 250, 300, 320])

# Consumption expenditure (Y)
consumption = np.array([25, 20, 40, 65, 120, 190, 200, 200, 260, 220])

# Create the regression design matrix
X = np.column_stack((np.ones(len(income)), income))

# Estimate regression coefficients using least squares
beta = np.linalg.solve(X.T @ X, X.T @ consumption)

# Display the coefficients
print("Regression coefficients:")
print(f"Intercept (beta_0): {beta[0]:.4f}")
print(f"Slope (beta_1): {beta[1]:.4f}")
```

```python
# Disposable income for the new household
new_income = 181

# Estimate consumption using the regression coefficients from Question 4(i)
predicted_consumption = beta[0] + beta[1] * new_income

# Display the predicted consumption
print("Disposable income:", new_income)
print(f"Estimated consumption: {predicted_consumption:.4f}")
```

```python
# Cross-check the design-matrix solution with NumPy's built-in least-squares solver
beta_check, *_ = np.linalg.lstsq(X, consumption, rcond=None)
print("Coefficients from normal equations:", beta)
print("Coefficients from np.linalg.lstsq: ", beta_check)
print("Results agree:", np.allclose(beta, beta_check))
```

**Conclusion**

Using the design matrix \(X\) (a column of ones and the disposable-income values), the least-squares coefficients were obtained from the normal equations \((X^TX)\beta=X^Ty\). The fitted model is

$$\hat{Y} = -5.5715 + 0.8234\,X$$

The slope of about 0.82 means that each additional unit of disposable income is associated with an increase of about 0.82 units of consumption (the marginal propensity to consume). For a household with disposable income of 181, the predicted consumption is approximately **143.47** units (about 143.5). The coefficients were cross-checked with `np.linalg.lstsq`, which gave the same values.
