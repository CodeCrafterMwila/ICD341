### Question 1: Generate 5,000 Exponential Samples Using Inverse Transform

**Objective:** To generate 5,000 random samples from an exponential distribution using the inverse transform sampling method with a rate parameter of 3.

The inverse transform method generates random samples from a specified probability distribution using uniformly distributed random numbers.

The cumulative distribution function of an exponential distribution is:

F(x) = 1 - exp(-λx)

The inverse transformation is:

X = -ln(1-U) / λ

Where:
- U is a uniformly distributed random variable between 0 and 1.
- λ is the rate parameter.

For this experiment, λ = 3 and the required number of samples is 5,000.

```python
import numpy as np
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Number of samples
n = 5000

# Rate parameter
lam = 3

# Generate uniform random numbers
u = np.random.uniform(0, 1, n)

# Display the first 10 values
print("First 10 uniform random numbers: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:14")
print(u[:10])
```

```python
# Apply inverse transform sampling
exponential_samples = -np.log(1 - u) / lam

# Display the first 10 exponential samples
print("First 10 exponential samples: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:5")
print(exponential_samples[:10])

# Display total number of samples
print("Total samples generated: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:9", len(exponential_samples))
```

```python
# Calculate descriptive statistics

print("Sample Mean: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:3", np.mean(exponential_samples))

print("Sample Variance: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:5", np.var(exponential_samples))

print("Sample Standard Deviation: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:7", np.std(exponential_samples))

print("Minimum Value: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:9", np.min(exponential_samples))

print("Maximum Value: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:11", np.max(exponential_samples))
```

```python
import matplotlib.pyplot as plt

# Plot histogram
plt.figure(figsize=(10, 5))

plt.hist(
    exponential_samples,
    bins=50,
    density=True,
    alpha=0.7,
    label="Simulated Exponential Samples"
)

plt.title("Exponential Distribution Using Inverse Transform")
plt.xlabel("Sample Values")
plt.ylabel("Density")

plt.legend()
plt.grid(True)

plt.show()
```

### Conclusion

The inverse transform sampling method was successfully implemented to generate 5,000 random samples from an exponential distribution with a rate parameter of 3.

Uniform random numbers were generated and transformed using the inverse cumulative distribution function.

The simulated sample mean was approximately 0.3333, which is close to the theoretical mean of 1/3.

The histogram displayed a positively skewed distribution, demonstrating the expected characteristics of an exponential distribution.

### Question 2: Sampling from a Categorical Distribution

**Objective:** To generate random samples from a categorical distribution with probabilities [0.1, 0.6, 0.3].

A categorical distribution is a discrete probability distribution that assigns probabilities to a finite number of categories.

In this experiment, there are three categories with probabilities:

P(X = 0) = 0.1
P(X = 1) = 0.6
P(X = 2) = 0.3

The probabilities sum to 1, satisfying the requirement for a valid probability distribution.

NumPy's random choice function will be used to generate samples from the distribution.

```python
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Define categories
categories = [0, 1, 2]

# Define probabilities
probabilities = [0.1, 0.6, 0.3]

# Display probabilities
print("Categories:", categories)
print("Probabilities:", probabilities)
print("Sum of probabilities:", sum(probabilities))
```

```python
# Number of samples
n_samples = 10000

# Generate samples from the categorical distribution
samples = np.random.choice(
    categories,
    size=n_samples,
    p=probabilities
)

# Display the first 20 samples
print("First 20 samples: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:12")
print(samples[:20])

print("Total samples generated: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:15", len(samples))
```

```python
# Calculate frequencies
values, counts = np.unique(samples, return_counts=True)

# Calculate relative frequencies
relative_frequencies = counts / n_samples

# Display results
print("Category Frequencies: - ICD341_Lab11_Mwila_Singoma_202401774.ipynb:8")

for value, count, relative_frequency in zip(
    values, counts, relative_frequencies
):
    print(
        f"Category {value}: Count = {count}, "
        f"Relative Frequency = {relative_frequency:.4f}"
    )
```

```python
import matplotlib.pyplot as plt

# Plot category frequencies
plt.figure(figsize=(8, 5))

plt.bar(values, relative_frequencies)

plt.axhline(
    y=0.1,
    linestyle="--",
    label="P(Category 0) = 0.1"
)

plt.axhline(
    y=0.6,
    linestyle="--",
    label="P(Category 1) = 0.6"
)

plt.axhline(
    y=0.3,
    linestyle="--",
    label="P(Category 2) = 0.3"
)

plt.title("Categorical Distribution Sampling")
plt.xlabel("Category")
plt.ylabel("Relative Frequency")
plt.xticks(categories)
plt.legend()
plt.grid(axis="y")

plt.show()
```

### Conclusion

A categorical distribution with probabilities [0.1, 0.6, 0.3] was successfully simulated using NumPy.

A total of 10,000 random samples were generated.

The observed relative frequencies were approximately 0.1, 0.6, and 0.3 for categories 0, 1, and 2, respectively.

The results demonstrate that the simulated samples follow the specified categorical probability distribution.

## Question 3: Rejection Sampling

Rejection sampling is a method for generating random samples from a target
probability distribution using a proposal distribution.

For this question, the target distribution is Beta(2, 2), whose probability
density function is:

f(x) = 6x(1-x), for 0 ≤ x ≤ 1.

We use a Uniform(0, 1) proposal distribution. Since the maximum value of
the Beta(2, 2) density is 1.5, the acceptance probability is:

f(x) / 1.5 = 4x(1-x).

A proposed value is accepted if a uniform random number is less than or
equal to this acceptance probability.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Number of samples required
n = 5000

# Store accepted samples
samples = []

# Count total proposals
proposals = 0

# Rejection sampling
while len(samples) < n:
    # Propose a value from Uniform(0, 1)
    x = np.random.uniform(0, 1)

    # Generate a uniform random number
    u = np.random.uniform(0, 1)

    # Acceptance probability for Beta(2, 2)
    acceptance_probability = 4 * x * (1 - x)

    proposals += 1

    # Accept or reject the proposed value
    if u <= acceptance_probability:
        samples.append(x)

# Convert samples to NumPy array
samples = np.array(samples)

# Display results
print("Number of accepted samples:", len(samples))
print("Total proposals:", proposals)
print("Sample mean:", np.mean(samples))
print("Sample variance:", np.var(samples))

# Plot histogram and theoretical density
x_values = np.linspace(0, 1, 200)
theoretical_density = 6 * x_values * (1 - x_values)

plt.hist(samples, bins=30, density=True, alpha=0.7, label="Generated samples")
plt.plot(x_values, theoretical_density, linewidth=2, label="Beta(2, 2) density")

plt.title("Rejection Sampling: Beta(2, 2)")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.show()
```

### Interpretation

Rejection sampling successfully generated 5,000 samples from the Beta(2, 2)
distribution using a Uniform(0, 1) proposal distribution.

The histogram is symmetric around 0.5, and most observations are concentrated
near the center of the interval. The sample mean is expected to be close to
the theoretical mean of 0.5.

This demonstrates how rejection sampling can generate samples from a target
distribution using a simpler proposal distribution.

## Question 4: Four Main Steps of a Particle Filter

A particle filter is a sequential Monte Carlo method used to estimate the
state of a system when the system is dynamic and observations are noisy.

The four main steps are:

### 1. Initialization
Generate a set of particles representing possible initial states of the
system. Assign each particle an initial weight, usually equal to 1/N,
where N is the number of particles.

### 2. Prediction
Move or propagate each particle forward using the system's transition model.
This predicts the possible states of the system at the next time step.

### 3. Weight Update
Compare each predicted particle with the new observation. Assign higher
weights to particles that are more consistent with the observation and
lower weights to particles that are less consistent.

### 4. Resampling
Select particles according to their weights. Particles with higher weights
are more likely to be selected, while particles with low weights are less
likely to survive. The selected particles form a new particle set for the
next time step.

## Question 5: Simulation-Based Calibration (SBC)

Simulation-Based Calibration (SBC) is a method used to check whether a
Bayesian inference algorithm produces calibrated posterior distributions.

The general procedure is:

1. Simulate parameter values from the prior distribution.
2. Generate simulated data using those parameter values.
3. Fit the Bayesian model to the simulated data.
4. Draw samples from the posterior distribution.
5. Calculate the rank of the true parameter value among the posterior samples.
6. Repeat the procedure many times and plot the rank histogram.

If the Bayesian inference algorithm is correctly calibrated, the rank
histogram should be approximately uniform.

A non-uniform rank histogram may indicate that the inference algorithm
is not properly calibrated. Possible causes include biased posterior
estimates, sampling problems, poor convergence, or errors in the model
implementation.

Common patterns include:

- U-shaped histogram: posterior distributions may be too narrow
  (under-dispersed).
- Hump-shaped histogram: posterior distributions may be too wide
  (over-dispersed).
- Skewed histogram: may indicate systematic bias in the estimates.

Therefore, a non-uniform SBC rank histogram suggests that the Bayesian
model or inference procedure should be investigated.
