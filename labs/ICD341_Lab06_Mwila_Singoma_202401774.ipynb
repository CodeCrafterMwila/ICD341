**QUESTION 1**: Create a covariance matrix and perform Cholesky decomposition.

## Question 1: Covariance Matrix and Cholesky Decomposition

**Objective:** To create a covariance matrix using NumPy and perform Cholesky decomposition.

A covariance matrix describes the variances of individual variables and the covariances between pairs of variables.

Cholesky decomposition expresses a symmetric, positive-definite matrix as the product of a lower-triangular matrix and its transpose:

$$
A = LL^T
$$

Where:

* \(A\) is the original matrix.
* \(L\) is a lower-triangular matrix.
* \(L^T\) is the transpose of \(L\).


```python
import numpy as np

# Create sample observations for two variables
data = np.array([
    [2, 5],
    [4, 4],
    [6, 8],
    [8, 7],
    [10, 13]
])

# Calculate the covariance matrix
cov_matrix = np.cov(data, rowvar=False)

# Perform Cholesky decomposition
L = np.linalg.cholesky(cov_matrix)

# Display the covariance matrix
print("Covariance Matrix:")
print(cov_matrix)

# Display the Cholesky factor
print("\nCholesky Decomposition (L):")
print(L)

# Verify the decomposition
print("\nReconstructed Covariance Matrix:")
print(L @ L.T)
```

**Conclusion:**

A covariance matrix was created using NumPy to calculate the variances and covariance of two variables. Cholesky decomposition was then performed to express the covariance matrix as the product of a lower-triangular matrix and its transpose. The decomposition was verified by reconstructing the original covariance matrix using matrix multiplication.


**QUESTION 2**: Perform QR decomposition on a matrix.


## Question 2: QR Decomposition

**Objective:** To perform QR decomposition on a matrix using NumPy.

QR decomposition expresses a matrix as the product of an orthogonal matrix \(Q\) and an upper-triangular matrix \(R\).

The decomposition is represented as:

$$
A=QR
$$

The decomposition can be verified by multiplying \(Q\) by \(R\) to reconstruct the original matrix.


```python
import numpy as np

# Create a matrix
A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

# Perform QR decomposition
Q, R = np.linalg.qr(A)

# Display the original matrix
print("Original Matrix A:")
print(A)

# Display the orthogonal matrix Q
print("\nOrthogonal Matrix Q:")
print(Q)

# Display the upper-triangular matrix R
print("\nUpper-Triangular Matrix R:")
print(R)

# Reconstruct the original matrix
print("\nReconstructed Matrix (Q @ R):")
print(Q @ R)
```

**Conclusion:**

QR decomposition was performed using NumPy to express the original matrix as the product of an orthogonal matrix \(Q\) and an upper-triangular matrix \(R\). The decomposition was verified by multiplying \(Q\) and \(R\), which reconstructed the original matrix. QR decomposition is useful in numerical linear algebra, particularly for solving least-squares problems and systems of linear equations.


**QUESTION 3**: Reconstruct a matrix using SVD.

## Question 3: Matrix Reconstruction Using SVD

**Objective:** To reconstruct a matrix using Singular Value Decomposition (SVD) in NumPy.

Singular Value Decomposition expresses a matrix as the product of three matrices:

$$
A=U\Sigma V^T
$$

The original matrix can be reconstructed by multiplying the left singular vector matrix, the diagonal matrix of singular values, and the transpose of the right singular vector matrix.


```python
import numpy as np

# Create the original matrix
A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

# Perform Singular Value Decomposition
U, S, VT = np.linalg.svd(A, full_matrices=False)

# Convert singular values into a diagonal matrix
Sigma = np.diag(S)

# Reconstruct the original matrix
A_reconstructed = U @ Sigma @ VT

# Display the original matrix
print("Original Matrix A:")
print(A)

# Display the singular values
print("\nSingular Values:")
print(S)

# Display the reconstructed matrix
print("\nReconstructed Matrix:")
print(A_reconstructed)

# Verify the reconstruction
print("\nReconstruction Successful:")
print(np.allclose(A, A_reconstructed))
```

**Conclusion:**

Singular Value Decomposition was performed using NumPy to decompose the original matrix into the matrices \(U\), \(\Sigma\), and \(V^T\). The original matrix was successfully reconstructed by multiplying these three components. The reconstruction was verified using `np.allclose()`, which returned `True`. SVD is useful in numerical analysis, dimensionality reduction, and data compression.

## Question 4: Calculate the Condition Number of a Matrix

**Objective:** To calculate the condition number of a matrix using NumPy.

The condition number measures the sensitivity of a matrix problem to small changes in the input data. A small condition number generally indicates that the matrix is well-conditioned, while a large condition number indicates that it may be ill-conditioned.

The condition number using the matrix 2-norm is:

$$
\kappa_2(A)=\frac{\sigma_{\max}}{\sigma_{\min}}
$$

where \(\sigma_{\max}\) and \(\sigma_{\min}\) are the largest and smallest singular values of the matrix, respectively.


```python
import numpy as np

# A well-conditioned matrix
A = np.array([
    [1, 2],
    [3, 4]
])

# A nearly singular (ill-conditioned) matrix for comparison
B = np.array([
    [1, 1],
    [1, 1.0001]
])

# Calculate the condition numbers
cond_A = np.linalg.cond(A)
cond_B = np.linalg.cond(B)

print("Matrix A:")
print(A)
print(f"Condition number of A: {cond_A:.4f}")

print("\nMatrix B:")
print(B)
print(f"Condition number of B: {cond_B:.4f}")
```

**Conclusion:**

The condition number was calculated using NumPy's `np.linalg.cond()` function. Matrix \(A\) has a condition number of about 14.93, which is small, so \(A\) is **well-conditioned**: small changes in the input produce only small changes in the solution of \(Ax=b\). Matrix \(B\) is nearly singular (its rows are almost identical) and has a condition number of about \(4\times10^{4}\), so it is **ill-conditioned**: tiny changes in the data can produce large changes in the solution. A condition number close to 1 is ideal, and very large values (for example \(10^{10}\) or more) warn that numerical results may be unreliable.

**QUESTION 5**: Create a sparse 1000 × 1000 matrix.

## Question 5: Create a Sparse 1000 × 1000 Matrix

**Objective:** To create a sparse matrix of size \(1000\times1000\) using SciPy.

A sparse matrix is a matrix in which most elements are zero. Sparse matrix representations store nonzero elements efficiently, reducing memory consumption when working with large matrices containing many zeros.


```python
import numpy as np
from scipy.sparse import random

# Create a sparse 1000 x 1000 matrix with 1% nonzero entries
A = random(
    1000,
    1000,
    density=0.01,
    format="csr",
    random_state=42
)

# Display matrix information
print("Matrix Shape:", A.shape)
print("Number of Nonzero Elements:", A.nnz)
print("Matrix Density:", A.nnz / (1000 * 1000))

# Display a few of the stored nonzero values
rows, cols = A.nonzero()
print("\nFirst 5 nonzero entries (row, column, value):")
for r, c in list(zip(rows, cols))[:5]:
    print(f"  ({r}, {c}) -> {A[r, c]:.4f}")

# Compare memory usage with a dense array of the same size
sparse_bytes = A.data.nbytes + A.indices.nbytes + A.indptr.nbytes
dense_bytes = A.toarray().nbytes
print(f"\nMemory used by sparse (CSR) matrix: {sparse_bytes / 1024:.1f} KB")
print(f"Memory used by dense matrix:        {dense_bytes / 1024:.1f} KB")
print(f"The dense matrix needs about {dense_bytes / sparse_bytes:.0f} times more memory.")
```

**Conclusion:**

A sparse \(1000\times1000\) matrix was created using SciPy with a density of 0.01, giving 10,000 nonzero entries out of one million. It was stored in Compressed Sparse Row (CSR) format, which keeps only the nonzero values and their positions. The memory comparison shows that the CSR matrix needs roughly 60 times less memory than the equivalent dense array.

**QUESTION 6**: Explain why sparse matrix are useful for large datasets.


## Question 6: Explain Why Sparse Matrices Are Useful for Large Datasets

Sparse matrices are useful for large datasets because they contain mostly zero values and allow these values to be stored efficiently.

The main advantages of sparse matrices are:

1. **Reduced memory usage:** Sparse matrices store nonzero values and their positions instead of storing every element, saving memory when most entries are zero.

2. **Faster computations:** Operations involving sparse matrices can be faster because computations involving zero elements can often be skipped.

3. **Efficient storage:** Sparse matrix formats, such as Compressed Sparse Row (CSR), store large matrices efficiently.

4. **Scalability:** Sparse matrices make it practical to work with very large datasets that would otherwise require substantial memory.

5. **Applications in data science:** Sparse matrices are useful in machine learning, recommendation systems, graph analysis, natural language processing, and scientific computing.

**Conclusion:**

Sparse matrices are important in advanced statistical computing because they reduce memory requirements and can improve computational efficiency when working with large datasets containing many zero values. Their use enables efficient processing of large-scale data.
