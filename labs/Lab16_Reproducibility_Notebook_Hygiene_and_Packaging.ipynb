## Exercise 1: Setting Random Seeds

A random seed is a value used to initialise a random-number generator.

Setting a random seed helps ensure reproducibility by allowing random operations to produce the same results when repeated under the same conditions.

In this exercise, we will set random seeds for Python, NumPy and PyTorch.

````python
import random
import numpy as np
import torch

print("All libraries imported successfully!")
````

````python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("Random seeds have been set successfully.")
````

````python
# Set the seed
np.random.seed(42)

# Generate the first sequence
values1 = np.random.rand(5)

print("First sequence:")
print(values1)
````

````python
# Reset the seed to the same value
np.random.seed(42)

# Generate the second sequence
values2 = np.random.rand(5)

print("Second sequence:")
print(values2)
````

````python
print("Are both sequences identical?")
print(np.array_equal(values1, values2))
````

## Exercise 2: Notebook Hygiene

Notebook hygiene involves following good practices when creating and maintaining Jupyter notebooks.

A well-organised notebook should be easy to read, understand, execute and reproduce.

Good notebook practices include:
- Using descriptive notebook and variable names.
- Organising code into logical sections.
- Including Markdown explanations.
- Keeping imports at the beginning.
- Avoiding unnecessary code duplication.
- Ensuring all cells execute without errors.
- Saving the completed notebook.

### Example: Organised Python Code

````python
# Generate sample maize production data
np.random.seed(123)

maize = np.random.normal(loc=5, scale=1.2, size=100)

# Calculate descriptive statistics
mean_production = np.mean(maize)
std_production = np.std(maize)

# Display results
print("Mean production:", mean_production)
print("Standard deviation:", std_production)
````

### Interpretation

The generated maize production dataset contains 100 observations, with a sample mean of approximately **4.98 tonnes** and a sample standard deviation of approximately **1.17 tonnes**.

The mean represents the average maize production across the observations, while the standard deviation measures how spread out the values are around the mean.

The results are close to the parameters used to generate the data (mean = 5 tonnes and standard deviation = 1.2 tonnes). Small differences are expected because the observations are randomly generated.

Setting the random seed to 123 ensures that the same dataset can be reproduced when the code is executed again under the same conditions.

## Exercise 4: Python Project Packaging

Python project packaging involves organising project files and dependencies
so that a project can be easily shared, installed and maintained.

A properly organised Python project should contain:

1. README.md – project documentation.
2. requirements.txt – a list of required Python libraries.
3. Notebooks – Jupyter notebooks containing experiments.
4. Source code – reusable Python modules.
5. Data – datasets used in the project.

Packaging improves project organisation, portability, maintainability
and collaboration.

### Installing Project Dependencies

Python dependencies can be installed automatically using pip and a requirements file.

From the ICD341 repository root, run:

```bash
python -m pip install -r labs/requirements_lab16.txt
```

This command reads the requirements file and installs all the listed libraries.

Using a requirements file makes project setup easier and improves portability because developers can install the required packages without installing each library separately.

### Saving the Dataset as CSV

The generated maize production observations are stored
in a CSV file for future analysis and reproducibility.

````python
from pathlib import Path
import pandas as pd

# Create a DataFrame from the generated maize observations
df = pd.DataFrame({
    "Maize_Production": maize
})

# Locate the project directory from the current working directory
current_dir = Path.cwd().resolve()

project_dir = next(
    (folder for folder in [current_dir, *current_dir.parents]
     if folder.name == "ICD341"),
    None
)

if project_dir is None:
    raise FileNotFoundError("Could not locate the ICD341 project directory.")

# Save the dataset
data_dir = project_dir / "Maize_Production_Project" / "data"
data_dir.mkdir(parents=True, exist_ok=True)

output_path = data_dir / "maize_data.csv"

df.to_csv(output_path, index=False)

print("Dataset successfully saved!")
print("Number of observations:", len(df))
print("\nFirst five observations:")
print(df.head())
````
