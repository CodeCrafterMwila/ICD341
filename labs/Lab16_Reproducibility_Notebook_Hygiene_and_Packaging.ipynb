## Exercise 1: Setting Random Seeds

A random seed is a value used to initialise a random-number generator.

Setting a random seed helps ensure reproducibility by allowing random operations to produce the same results when repeated under the same conditions.

In this exercise, we will set random seeds for Python, NumPy and PyTorch.

```python
import random
import numpy as np
import torch

print("All libraries imported successfully!")
```

```python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("Random seeds have been set successfully. - Lab16_Reproducibility_Notebook_Hygiene_and_Packaging.ipynb:10")
```

```python
# Set the seed
np.random.seed(42)

# Generate the first sequence
values1 = np.random.rand(5)

print("First sequence: - Lab16_Reproducibility_Notebook_Hygiene_and_Packaging.ipynb:7")
print(values1)
```

```python
# Reset the seed to the same value
np.random.seed(42)

# Generate the second sequence
values2 = np.random.rand(5)

print("Second sequence: - Lab16_Reproducibility_Notebook_Hygiene_and_Packaging.ipynb:7")
print(values2)
```

```python
print("Are both sequences identical? - Lab16_Reproducibility_Notebook_Hygiene_and_Packaging.ipynb:1")
print(np.array_equal(values1, values2))
```

## Exercise 2: Notebook Hygiene

Notebook hygiene involves following good practices when creating and maintaining Jupyter notebooks.

A well-organised notebook should be easy to read, understand, execute and reproduce.

Good notebook practices include:
- Using descriptive notebook and variable names.
- Organising code into logical sections.
- Including Markdown explanations.
- Keeping imports at the beginning.
- Avoiding unnecessary code duplication.
- Ensuring all cells execute without errors.
- Saving the completed notebook.

### Example: Organised Python Code

```python
# 1. Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 2. Set random seed
np.random.seed(123)

# 3. Generate sample data
maize = np.random.normal(loc=5, scale=1.2, size=100)

# 4. Calculate descriptive statistics
mean_production = np.mean(maize)
std_production = np.std(maize)

# 5. Display results
print("Mean production:", mean_production)
print("Standard deviation:", std_production)
```

### Interpretation

The generated dataset has a mean of approximately 5.03 and a standard deviation of approximately 1.35.

The mean indicates the average value of the generated observations, while the standard deviation measures how spread out the observations are around the mean.

The results demonstrate how organised code makes statistical calculations easier to understand, reproduce and maintain.

## Exercise 4: Python Project Packaging

Python project packaging involves organising project files and dependencies
so that a project can be easily shared, installed and maintained.

A properly organised Python project should contain:

1. README.md – project documentation.
2. requirements.txt – a list of required Python libraries.
3. Notebooks – Jupyter notebooks containing experiments.
4. Source code – reusable Python modules.
5. Data – datasets used in the project.

Packaging improves project organisation, portability, maintainability
and collaboration.

### Installing Project Dependencies

Python dependencies can be installed automatically using pip and a requirements file.

The command is:

python -m pip install -r requirements_lab16.txt

This command reads the requirements file and installs all listed libraries.

This makes project setup easier and improves portability because developers
can install the required packages without installing each library separately.
