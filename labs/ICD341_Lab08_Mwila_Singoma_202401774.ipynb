### Question 1: Generate 10,000 Observations from a Normal Distribution

**Objective:** To simulate 10,000 observations from a normal distribution with a population mean of 50 and a population variance of 100.

A normal distribution is defined by its mean, \(\mu\), and standard deviation, \(\sigma\). The given population mean is 50, while the population variance is \(10^2=100\). Therefore, the standard deviation is 10.

Python's NumPy library provides the `np.random.normal()` function for generating random observations from a normal distribution. The simulated data will be analysed using descriptive statistics, including the sample mean, sample standard deviation, minimum, and maximum.


```python
# Import NumPy
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Generate 10,000 observations from N(50, 10²)
data = np.random.normal(loc=50, scale=10, size=10000)

# Display the results
print("Number of observations:", len(data))
print("Sample mean:", np.mean(data))
print("Sample standard deviation:", np.std(data))
print("Minimum value:", np.min(data))
print("Maximum value:", np.max(data))
```

**Conclusion**

A total of 10,000 random observations were generated from a normal distribution with a population mean of 50 and a population variance of 100. The sample mean was approximately 49.9786, while the sample standard deviation was approximately 10.0341.

These results are close to the specified population parameters. The small differences are due to random sampling. Therefore, the simulation successfully demonstrates how a sample generated from a normal distribution approximates its population characteristics.

### Question 2: Bernoulli Simulation

**Objective:** To generate 10,000 Bernoulli observations with a success probability of 0.3 and calculate the sample proportion of successes.

A Bernoulli random variable takes the value 1 with probability \(p\) and the value 0 with probability \(1-p\). In this experiment, the probability of success is 0.3, while the probability of failure is 0.7.

The sample proportion is calculated by dividing the number of successes by the total number of observations. As the sample size increases, the sample proportion is expected to approach the true probability of 0.3.


```python
# Import NumPy
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Generate 10,000 Bernoulli observations
data = np.random.binomial(n=1, p=0.3, size=10000)

# Calculate the number of successes
successes = np.sum(data)

# Calculate the sample proportion
sample_proportion = np.mean(data)

# Display the results
print("Number of observations:", len(data))
print("Number of successes:", successes)
print("Number of failures:", len(data) - successes)
print("True probability:", 0.3)
print("Sample proportion:", sample_proportion)

# Standard error of the sample proportion and how far the estimate is from p = 0.3
standard_error = np.sqrt(0.3 * 0.7 / len(data))
z_score = (sample_proportion - 0.3) / standard_error
print("Standard error:", standard_error)
print("Distance from 0.3 in standard errors:", z_score)
```

**Conclusion**

A total of 10,000 Bernoulli observations were generated with a success probability of 0.3. The simulation produced 2,887 successes and 7,113 failures, giving a sample proportion of 0.2887 (28.87%).

The theoretical standard error of a sample proportion with \(n=10{,}000\) and \(p=0.3\) is about 0.0046, so the estimate lies roughly 2.5 standard errors below 0.3. This is a somewhat large deviation for a single run, but it is still possible by chance and does not indicate a problem with the simulation. By the law of large numbers, the sample proportion approaches the true probability of 0.3 as the number of observations increases.

**QUESTION 3**: Use Monte Carlo simulation to estimate \int_0^1 x^3\,dx

### Question 3: Monte Carlo Estimation of an Integral

**Objective:** To estimate the integral \(\int_0^1 x^3\,dx\) using the Monte Carlo method.

The exact value of the integral can be calculated analytically:

$$
\int_0^1 x^3\,dx=\left[\frac{x^4}{4}\right]_0^1=0.25
$$

The Monte Carlo method estimates an integral by generating random observations over the integration interval and calculating the average function value.

For \(N\) uniformly distributed observations between 0 and 1, the integral is estimated using:

$$
I\approx\frac{1}{N}\sum_{i=1}^{N}X_i^3
$$

In this experiment, 100,000 random observations will be generated. The estimated value will then be compared with the exact integral of 0.25 to evaluate the accuracy of the Monte Carlo method.


```python
# Import NumPy
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Generate 100,000 random observations between 0 and 1
N = 100000
x = np.random.uniform(0, 1, N)

# Estimate the integral using the Monte Carlo method
estimated_integral = np.mean(x**3)

# Calculate the exact integral
exact_integral = 0.25

# Calculate the absolute error
absolute_error = abs(estimated_integral - exact_integral)

# Display the results
print("Number of observations:", N)
print("Monte Carlo estimate:", estimated_integral)
print("Exact integral:", exact_integral)
print("Absolute error:", absolute_error)
```

**Conclusion**

The Monte Carlo method was used to estimate the integral \(\int_0^1 x^3\,dx\) using 100,000 uniformly distributed random observations. The estimated integral was 0.24918, compared with the exact value of 0.25. The absolute error was approximately 0.0008224.

The small difference between the estimated and exact values demonstrates that Monte Carlo integration can provide a good approximation to a definite integral. Increasing the number of random observations will generally improve the estimate's precision.


**QUESTION 4**: Estimate 𝜋 using one million points.

### Question 4: Estimate Pi Using One Million Random Points

**Objective:** To estimate the mathematical constant \(\pi\) using the Monte Carlo method and one million randomly generated points.

Consider a circle of radius 1 enclosed within a square of side length 2. The circle has area \(\pi\), while the square has area 4. Consequently, the proportion of uniformly distributed random points falling inside the circle approximates \(\pi/4\).

The value of \(\pi\) can therefore be estimated using:

$$
\pi\approx4\left(\frac{\text{Number of points inside the circle}}{\text{Total number of points}}\right)
$$

A point lies inside the circle when \(x^2+y^2\leq1\). In this experiment, one million random points will be generated, and the resulting estimate will be compared with NumPy's value of \(\pi\).


```python
# Import NumPy
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Generate one million random points
N = 1_000_000
x = np.random.uniform(-1, 1, N)
y = np.random.uniform(-1, 1, N)

# Identify points inside or on the circle
inside_circle = (x**2 + y**2) <= 1

# Count the points inside the circle
number_inside = np.sum(inside_circle)

# Estimate pi
pi_estimate = 4 * number_inside / N

# Calculate the absolute error
absolute_error = abs(pi_estimate - np.pi)

# Display the results
print("Total number of points:", N)
print("Points inside the circle:", number_inside)
print("Estimated value of pi:", pi_estimate)
print("Actual value of pi:", np.pi)
print("Absolute error:", absolute_error)
```

**Conclusion**

The Monte Carlo method was used to estimate the value of \(\pi\) using one million randomly generated points. Of these points, 784,973 fell inside or on the boundary of a circle of radius 1. The estimated value of \(\pi\) was 3.139892, compared with the reference value of approximately 3.141593. The absolute error was approximately 0.001701.

The experiment demonstrates how Monte Carlo simulation can estimate mathematical constants using random sampling. The estimate may become more accurate as the number of random points increases, although individual simulations can vary.


**QUESTION 5**: Estimate 𝑃(𝑋 > 1.96) where 𝑋~𝑁(0, 1).

### Question 5: Estimate \(P(X>1.96)\)

**Objective:** To estimate the probability that a standard normal random variable exceeds 1.96 using Monte Carlo simulation.

A standard normal random variable has a mean of 0 and a standard deviation of 1. The required probability is \(P(X>1.96)\), representing the proportion of observations greater than 1.96.

The Monte Carlo estimate is calculated by dividing the number of simulated observations exceeding 1.96 by the total number of observations:

$$
\hat{P}(X>1.96)=\frac{\sum_{i=1}^{N}I(X_i>1.96)}{N}
$$

Here, \(I(X_i>1.96)\) equals 1 when an observation exceeds 1.96 and 0 otherwise.

The theoretical probability is approximately 0.025. A simulation with 1,000,000 observations will be used to estimate this probability and compare it with the theoretical value.


```python
# Import NumPy
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Generate one million observations from a standard normal distribution
N = 1_000_000
data = np.random.normal(loc=0, scale=1, size=N)

# Count observations greater than 1.96
count = np.sum(data > 1.96)

# Estimate the probability
estimated_probability = count / N

# Theoretical probability
theoretical_probability = 1 - 0.975

# Calculate the absolute error
absolute_error = abs(estimated_probability - theoretical_probability)

# Display the results
print("Number of observations:", N)
print("Observations greater than 1.96:", count)
print("Estimated probability:", estimated_probability)
print("Theoretical probability:", theoretical_probability)
print("Absolute error:", absolute_error)
```

**Conclusion**

Monte Carlo simulation was used to estimate the probability that a standard normal random variable exceeds 1.96. One million observations were generated, and 24,914 observations exceeded the specified threshold. The estimated probability was 0.024914, or 2.4914%, compared with the theoretical probability of approximately 0.025, or 2.5%.

The absolute error was 0.000086, demonstrating that the simulation produced an accurate approximation of the theoretical probability. This experiment illustrates how Monte Carlo methods can estimate probabilities by generating random samples and calculating the proportion of observations satisfying a specified condition.

### Question 6: Estimate \(E(X^2)\) Using Importance Sampling

**Objective:** To estimate the expectation \(E(X^2)\), where \(X\sim U(0,1)\), using importance sampling.

The exact expectation is:

$$
E(X^2)=\int_0^1x^2\,dx=\frac13\approx0.333333
$$

Importance sampling estimates an expectation by generating observations from a proposal distribution \(q(x)\) and weighting them according to the ratio of the target density \(f(x)\) to the proposal density.

The importance-sampling estimator is:

$$
\hat{I}=\frac1N\sum_{i=1}^{N}g(X_i)\frac{f(X_i)}{q(X_i)}
$$

For this experiment, the target density is \(f(x)=1\) on \([0,1]\), and the proposal density is \(q(x)=2x\) on the same interval. Therefore, the importance weight is \(1/(2x)\).

The estimate will be compared with the exact expectation of \(1/3\) to evaluate the accuracy of the method.


```python
# Import NumPy
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Number of observations
N = 100000

# Generate samples from q(x) = 2x on [0, 1]
# If U ~ Uniform(0, 1), then X = sqrt(U) has density q(x) = 2x
u = np.random.uniform(0, 1, N)
x = np.sqrt(u)

# Target density f(x) = 1
f_x = np.ones(N)

# Importance density q(x) = 2x
q_x = 2 * x

# Calculate importance weights
weights = f_x / q_x

# Estimate E(X^2) using importance sampling
estimated_expectation = np.mean((x**2) * weights)

# Calculate the exact expectation
exact_expectation = 1 / 3

# Calculate absolute error
absolute_error = abs(estimated_expectation - exact_expectation)

# Display results
print("Number of observations:", N)
print("Importance sampling estimate:", estimated_expectation)
print("Exact expectation:", exact_expectation)
print("Absolute error:", absolute_error)
```

**Conclusion**

Importance sampling was used to estimate the expected value of \(X^2\), where \(X\sim U(0,1)\). Using 100,000 simulated observations drawn from the proposal density \(q(x)=2x\), the estimated expectation was 0.333174. The exact expectation was \(1/3\), approximately 0.333333, giving an absolute error of approximately 0.0001595.

The estimated value is close to the exact expectation, demonstrating how importance sampling uses an alternative sampling distribution and appropriate weights to estimate an expected value.


**QUESTION 7**: Explain the difference between importance sampling and control variates.

### Question 7: Difference Between Importance Sampling and Control Variates

Importance sampling and control variates are Monte Carlo variance-reduction techniques used to improve the accuracy and efficiency of simulation-based estimates. However, they achieve this in different ways.

**1. Importance Sampling**

Importance sampling involves drawing observations from an alternative probability distribution called the proposal distribution. The observations are then assigned weights based on the ratio between the target probability density and the proposal density.

Its general estimator is:

$$
\hat{\mu}_{IS}=\frac{1}{N}\sum_{i=1}^{N}g(X_i)\frac{f(X_i)}{q(X_i)}
$$

Importance sampling is particularly useful when important or rare events occur infrequently under the original distribution. By choosing an appropriate proposal distribution, these events can be sampled more effectively.

**2. Control Variates**

Control variates reduce the variance of an estimator by using an additional variable that is correlated with the variable of interest and whose expected value is known.

A control-variate estimator can be written as:

$$
\hat{\mu}_{CV}=\bar{Y}-c(\bar{Z}-E[Z])
$$

Here, \(\bar{Y}\) is the original estimate, \(\bar{Z}\) is the sample mean of the control variable, \(E[Z]\) is its known expected value, and \(c\) is a coefficient used to reduce variance.

Control variates are most useful when a strongly correlated variable with a known expectation is available.

**3. Key Differences**

| Feature                    | Importance Sampling                                                    | Control Variates                                                                                            |
| -------------------------- | ---------------------------------------------------------------------- | ----------------------------------------------------------------------------------------------------------- |
| Main technique             | Changes the sampling distribution.                                     | Uses a correlated auxiliary variable.                                                                       |
| Requirement                | A suitable proposal distribution and correct importance weights.       | A correlated control variable with a known expected value.                                                  |
| How it improves estimation | Samples important regions more effectively and corrects using weights. | Adjusts the estimate using the difference between the control variable's sample mean and known expectation. |
| Typical application        | Estimating rare-event probabilities and expectations.                  | Improving estimates in simulations where useful correlated variables are available.                         |

**Conclusion**

Importance sampling improves Monte Carlo estimation by changing the sampling distribution and applying appropriate weights. Control variates improve estimation by using a correlated variable with a known expected value to reduce variance. Both methods can improve simulation efficiency, but their effectiveness depends on choosing an appropriate proposal distribution or control variable.
